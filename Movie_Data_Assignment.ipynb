### **Question 23**

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
# Gross revenue ko numeric form me convert krna
df["Gross"] = pd.to_numeric(
    df['Gross'].astype(str).str.replace(',', '', regex=False),
    errors='coerce'
)

# Drama aur Action movies filter karna

drama = df[
    df['Genre'].str.contains('Drama', na=False) &
    ~df['Genre'].str.contains('Action', na=False)
]['Gross'].dropna()

action = df[
    df['Genre'].str.contains('Action', na=False) &
    ~df['Genre'].str.contains('Drama', na=False)
]['Gross'].dropna()

# Violion plot banana

plt.figure(figsize=(12, 7))

plt.violinplot(
    [drama, action],
    showmedians=True
)

plt.xticks([1, 2], ['Drama', 'Action'])
plt.title('Comparison of Gross Revenue: Drama vs Action Movies')
plt.xlabel('Genre')
plt.ylabel('Gross Revenue')
plt.ticklabel_format(axis='y', style='plain')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# IMDB rating ke mutabiq Top 5 movies select karna
top5 = df.nlargest(5, 'IMDB_Rating').copy()

# IMDB rating ko 100 scale par convert karna
top5['IMDB_Scaled'] = top5['IMDB_Rating'] * 10

# Missing Meta_score wali movies remove karna
top5 = top5.dropna(subset=['Meta_score'])

# Bar chart ki positions
x = np.arange(len(top5))
width = 0.35

# Chart banana
plt.figure(figsize=(14, 7))

bars1 = plt.bar(
    x - width/2,
    top5['Meta_score'],
    width,
    label='Meta_score',
    edgecolor='black'
)

bars2 = plt.bar(
    x + width/2,
    top5['IMDB_Scaled'],
    width,
    label='IMDB_Rating (Scaled to 100)',
    edgecolor='black'
)

# Bars par values show karna
plt.bar_label(bars1, fmt='%.1f', padding=3)
plt.bar_label(bars2, fmt='%.1f', padding=3)

# Titles aur labels
plt.title(
    'Top 5 Highest Rated Movies: Meta_score vs IMDB_Rating (Scaled to 100)',
    fontsize=14
)
plt.xlabel('Movie')
plt.ylabel('Score (out of 100)')

# Movie names display karna
labels = [
    f"{title}\n({year})"
    for title, year in zip(top5['Series_Title'], top5['Released_Year'])
]
plt.xticks(x, labels, rotation=0, ha='center')

plt.ylim(0, 110)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```
